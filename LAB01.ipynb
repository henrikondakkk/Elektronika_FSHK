## 2. Baza Teorike 

### Baza Teorike

Ideja kryesore pas kesaj gjuhe eshte krijimi i nje formati teksti te thjeshte qe te jete lehtesisht i lexueshem nga njerezit ne formen e tij origjinale, por njekohesisht i konvertueshem me lehtesi ne HTML te vlefshme.

### Disa nga Perdorimet Kryesore

Gjate kerkimit tim vura re se Markdown perdoret ne :

* **Dokumentimin e softuerit:** Skedaret `README.md` ne platforma si GitHub.

* **Publikimet ne web:** Perdorimi i Markdown per shkrimin e artikujve.

### Elementet Kryesore te Sintakses

Gjate eksperimentimit tim, identifikova elementet baze qe perdor me shpesh:

* **Titujt (Headings):** Perdorimi i simbolit `#` (nga `#` deri ne `######`).

* **Stili i tekstit:** Tekst *italic* me `*tekst*`, **bold** me `**tekst**`, dhe strikethrough me `~~tekst~~`.

* **Listat:** Listat e parenditura krijohen me `-` ose `*`, ndersa ato te renditura me numra `1.`, `2.`.

* **Lidhjet dhe Imazhet:**

  * Lidhje: `[Titulli](URL)`

  * Imazh: `![Alt Text](URL_e_imazhit)`

* **Blloku i Kodit:** Perdorimi i backticks (\`\`\`) per te shfaqur kod kompjuterik me sintakse te ngjyrosur.

## 3. Veprimet e Kryera ne Markdown

Gjate kesaj seance laboratorike une bera disa veprime praktike per te testuar funksionin e Markdown-it:

 **Strukturimi i Dokumentit:** Ndertova nje dokument nga e para duke perdorur tituj te niveleve te ndryshme (`#`, `##`, `###`) per te krijuar nje renditje te qarte .

**Krijimi i Tabelave:** Perdora simbolin `|` per ndarjen e kolonave dhe `-` per ndarjen e rreshtave.

   *Shembull nga tabela qe ndertova:*

   | Elementi | Kodi | Rezultati  | 
   | ----- | ----- | ----- | 
   | Bold | `**Tekst**` | **Tekst** | 
   | Italic | `*Tekst*` | *Tekst* | 
   | Code Inline | \`kod\` | `kod` | 

 **Integrimi i Formative Matematikore:** Perdora sintaksen LaTeX te mbeshtetur nga shume redaktore Markdown per te shkruar ekuacione si $E = mc^2$ apo formula me komplekse:
   

   $$
   f(x) = \int_{-\infty}^{\infty} e^{-x^2} dx
   $$


## 4. Analiza e Markdown-it

Nga testimet e mia praktike , une dola ne perparsit e te perdorimit te kesaj gjuhe

### Perparsite:

* **Shpejtesia :** Nuk kam nevoj te heq duart nga tastiera per te perdorur mausin kur dua te formatoj tekstin.

* **Portabiliteti:** Skedaret `.md` jane tekste te thjeshta dhe mund te hapen ne çdo sistem operativ (Windows, macOS, Linux) me cdo redaktor teksti.

* **Sistemimi:** Jane ideal per tu perdorur me sisteme kontrolli versioni si Git, sepse ndryshimet ne tekst mund te gjurmohen linje per linj.
